```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Optional: make plots look cleaner
sns.set_style("whitegrid")
```

```python
titanic = sns.load_dataset("titanic")

titanic.head()
```

```python
titanic["survived"].value_counts()
```

```python
sns.countplot(data=titanic, x="survived")

plt.title("Number of Survivors and Non-Survivors")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.countplot(data=titanic, x="sex")

plt.title("Number of Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
titanic["who"].value_counts()
```

```python
titanic["who"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)

plt.title("Passenger Categories")
plt.ylabel("")
plt.show()
```

```python
titanic["embark_town"].nunique()
```

```python
titanic["embark_town"].value_counts()
```

```python
titanic["embark_town"].value_counts().plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Passengers by Embark Town")
plt.xlabel("Embark Town")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=0)
plt.show()
```

```python
titanic.isnull().sum()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    titanic.isnull(),
    yticklabels=False,
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values in Titanic Dataset")
plt.show()
```

```python
sns.countplot(data=titanic, x="survived", palette="autumn", hue="sex")
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=titanic,
    x="survived",
    hue="sex",
    palette="autumn"
)

plt.title("Survival by Gender")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=titanic,
    x="survived",
    hue="class",
    palette="viridis"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

titanic["age"].plot(
    kind="hist",
    bins=30
)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="class"
)

plt.title("Age vs Fare by Passenger Class")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex"
)

plt.title("Age vs Fare by Gender")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
numeric_data = titanic.select_dtypes(include="number")

numeric_data.head()
```

```python
correlation = numeric_data.corr()

correlation
```

```python
correlation["survived"]
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="crest",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

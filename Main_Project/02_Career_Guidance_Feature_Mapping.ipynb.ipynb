
## 6. Draft Question-to-Feature Mapping

This section creates an initial mapping specification for all 120 questions. It records the question category, response type, proposed feature role, and mapping status.

The mapping is a draft. Scoring rules and target-label alignment must be validated before the specification is used to transform real user responses or train the final recommendation model.


```python

# Create a draft mapping table without modifying the source questionnaire

def classify_question(row):
    qid = row["Question_ID"]
    section = row["Section"]
    answer_type = row["Answer_Type"]

    if qid <= "Q010" and section == "Interest":
        role = "Assessment feature"
        status = "Needs semantic feature mapping"

    elif section == "Aptitude":
        role = "Assessment feature"
        status = "Needs answer-key and scoring validation"

    elif section == "Personality":
        role = "Assessment feature"
        status = "Needs trait scoring validation"

    elif section == "Skills":
        role = "Assessment feature"
        status = "Needs skill scoring and N/A handling"

    elif qid >= "Q061":
        role = "Career preference"
        status = "Needs domain-preference encoding"

    else:
        role = "Context or constraint"
        status = "Needs encoding and recommendation-role review"

    return pd.Series({
        "Proposed_Role": role,
        "Mapping_Status": status,
        "Answer_Type": answer_type
    })


mapping_df = questions_df.copy()

classification = mapping_df.apply(classify_question, axis=1)
mapping_df["Proposed_Role"] = classification["Proposed_Role"]
mapping_df["Mapping_Status"] = classification["Mapping_Status"]

# Add empty columns for decisions that require further validation
mapping_df["Proposed_Feature"] = pd.NA
mapping_df["Proposed_Scoring_Method"] = pd.NA
mapping_df["Encoding_Method"] = pd.NA

# Display a summary of the draft
print("Draft mapping created.")
print("Total questions:", len(mapping_df))

print("\nQuestions by proposed role:")
display(
    mapping_df["Proposed_Role"]
    .value_counts()
    .rename_axis("Proposed_Role")
    .reset_index(name="Question_Count")
)

print("\nSample mapping rows:")
display(
    mapping_df[
        [
            "Question_ID",
            "Section",
            "Answer_Type",
            "Proposed_Role",
            "Proposed_Feature",
            "Proposed_Scoring_Method",
            "Mapping_Status"
        ]
    ].head(20)
)
```

```python

# Step 8D: Map Q041-Q120 to profile and career-preference features

import pandas as pd
import re

# Ensure question IDs are consistent
questions_df["Question_ID"] = questions_df["Question_ID"].astype(str).str.strip()

# Feature names for Q041-Q060: profile and context
profile_features = {
    "Q041": "profile_education_level",
    "Q042": "profile_academic_field",
    "Q043": "profile_academic_performance",
    "Q044": "profile_certification_status",
    "Q045": "profile_training_duration",
    "Q046": "profile_work_experience",
    "Q047": "profile_internship_experience",
    "Q048": "profile_leadership_experience",
    "Q049": "profile_freelance_business_experience",
    "Q050": "profile_team_size_experience",
    "Q051": "profile_primary_career_motivation",
    "Q052": "profile_minimum_salary_expectation",
    "Q053": "profile_relocation_willingness",
    "Q054": "profile_five_year_goal",
    "Q055": "profile_stability_growth_preference",
    "Q056": "profile_work_environment",
    "Q057": "profile_company_size_preference",
    "Q058": "profile_travel_willingness",
    "Q059": "profile_work_schedule_preference",
    "Q060": "profile_performance_evaluation_preference",
}

# Map domain names from the questionnaire's Purpose column
domain_slug_map = {
    "Software Development": "software_development",
    "Data Science": "data_science",
    "Cybersecurity": "cybersecurity",
    "UI/UX Design": "ui_ux_design",
    "Business & Management": "business_management",
    "Healthcare": "healthcare",
    "Education & Research": "education_research",
    "Digital Marketing": "digital_marketing",
    "Finance & Accounting": "finance_accounting",
    "Engineering & Tech": "engineering_technology",
    "Creative Arts & Media": "creative_arts_media",
    "Law & Public Service": "law_public_service",
}

mapping_records = []

for _, row in questions_df.iterrows():
    qid = row["Question_ID"]
    match = re.fullmatch(r"Q(\d{3})", qid)

    if not match:
        continue

    q_num = int(match.group(1))

    if 41 <= q_num <= 60:
        feature = profile_features[qid]
        scoring = (
            "Preserve selected category; encode categorically. "
            "Use ordinal encoding only when option order is meaningful."
        )
        status = "Proposed profile feature; validate encoding"

    elif 61 <= q_num <= 120:
        purpose = str(row.get("Purpose", "")).strip()
        domain = purpose.replace("Domain Preference:", "").strip()
        slug = domain_slug_map.get(domain)

        if slug is None:
            feature = None
            scoring = "Review Purpose and define domain mapping"
            status = "Needs review"
        else:
            feature = f"career_pref_{slug}_{qid.lower()}"
            scoring = (
                "Preserve selected option; one-hot encode nominal options "
                "or use justified ordinal encoding. Define domain weights later."
            )
            status = "Proposed domain-preference feature; validate scoring"

    else:
        continue

    mapping_records.append({
        "Question_ID": qid,
        "Section": row.get("Section"),
        "Question_Text": row.get("Question_Text"),
        "Answer_Type": row.get("Answer_Type"),
        "Answer_Options": row.get("Answer_Options"),
        "Purpose": row.get("Purpose"),
        "Proposed_Feature": feature,
        "Proposed_Scoring_Method": scoring,
        "Mapping_Status": status,
    })

mapping_80 = pd.DataFrame(mapping_records)

# Sort numerically by question ID
mapping_80["_number"] = mapping_80["Question_ID"].str[1:].astype(int)
mapping_80 = mapping_80.sort_values("_number").drop(columns="_number")

display(mapping_80)

print("Questions mapped:", len(mapping_80))
print("Missing proposed features:", mapping_80["Proposed_Feature"].isna().sum())
print("Profile questions:", mapping_80["Question_ID"].str[1:].astype(int).between(41, 60).sum())
print("Domain-preference questions:", mapping_80["Question_ID"].str[1:].astype(int).between(61, 120).sum())
```

## Step 10F — Verify the Feature Scale and Dataset Lineage

```python
# ================================================================
# STEP 10F: SCALE AND DATASET LINEAGE VERIFICATION
# ================================================================

from pathlib import Path
import numpy as np
import pandas as pd

# ------------------------------------------------
# 1. Define dataset directory
# ------------------------------------------------

BASE_DIR = Path(
    r"C:\B-TECH\ACADEMIC\MLLAB\ML_PROJECT"
    r"\AI_Career_Guidance_Dataset_Package"
    r"\AI_Career_Guidance_Dataset"
)

# ------------------------------------------------
# 2. Load datasets
# ------------------------------------------------

processed = pd.read_csv(BASE_DIR / "career_guidance_processed.csv")
train = pd.read_csv(BASE_DIR / "career_guidance_train.csv")
test = pd.read_csv(BASE_DIR / "career_guidance_test.csv")

print("Datasets loaded successfully.")
print("Processed dataset:", processed.shape)
print("Training dataset:", train.shape)
print("Testing dataset:", test.shape)

# ------------------------------------------------
# 3. Define core features and target
# ------------------------------------------------

core_features = [
    "interest_technology",
    "interest_data",
    "interest_problem_solving",
    "interest_creativity",
    "interest_people",
    "interest_helping",
    "aptitude_logical",
    "aptitude_numerical",
    "aptitude_verbal",
    "aptitude_spatial",
    "aptitude_critical_thinking",
    "trait_openness",
    "trait_conscientiousness",
    "trait_extraversion",
    "trait_agreeableness",
    "trait_emotional_stability",
    "skill_programming",
    "skill_data_analysis",
    "skill_research",
    "skill_communication",
    "skill_leadership"
]

target = "recommended_career_domain"
columns = core_features + [target]

# Check that required columns exist
for dataset_name, df in [
    ("Processed", processed),
    ("Training", train),
    ("Testing", test)
]:
    missing_columns = [col for col in columns if col not in df.columns]

    if missing_columns:
        raise ValueError(
            f"{dataset_name} dataset is missing columns: "
            f"{missing_columns}"
        )

print("\nAll required features and target columns are present.")

# ------------------------------------------------
# 4. Verify original feature scales
# ------------------------------------------------

print("\n" + "=" * 65)
print("ORIGINAL FEATURE SCALE")
print("=" * 65)

print("\nProcessed dataset feature range:")
print("Minimum:", processed[core_features].min().min())
print("Maximum:", processed[core_features].max().max())

print("\nCombined training/testing feature range:")
combined = pd.concat(
    [train[columns], test[columns]],
    ignore_index=True
)

print("Minimum:", combined[core_features].min().min())
print("Maximum:", combined[core_features].max().max())

# ------------------------------------------------
# 5. Reverse-transform processed features
# ------------------------------------------------
# Candidate transformation:
# Original scale = 1 + 9 * normalized value
#
# This converts a normalized scale of 0–1
# to a candidate scale of 1–10.

reconstructed = processed[columns].copy()

reconstructed[core_features] = (
    1 + 9 * reconstructed[core_features]
)

# ------------------------------------------------
# 6. Compare feature means
# ------------------------------------------------

print("\n" + "=" * 65)
print("FEATURE MEAN COMPARISON")
print("=" * 65)

processed_means = reconstructed[core_features].mean()
combined_means = combined[core_features].mean()

mean_comparison = pd.DataFrame({
    "Reconstructed_Processed_Mean": processed_means,
    "Combined_Train_Test_Mean": combined_means
})

mean_comparison["Absolute_Mean_Difference"] = (
    mean_comparison["Reconstructed_Processed_Mean"]
    - mean_comparison["Combined_Train_Test_Mean"]
).abs()

print(mean_comparison.round(6))

print(
    "\nMaximum absolute difference between feature means:",
    round(mean_comparison["Absolute_Mean_Difference"].max(), 6)
)

# ------------------------------------------------
# 7. Verify complete record lineage
# ------------------------------------------------

def make_fingerprints(df):
    """
    Convert each complete feature-and-target row into a tuple.
    Round feature values to six decimal places to reduce
    insignificant floating-point differences.
    """

    temp = df[columns].copy()

    temp[core_features] = temp[core_features].round(6)

    return pd.Series(
        [
            tuple(row)
            for row in temp.itertuples(
                index=False,
                name=None
            )
        ],
        dtype="object"
    )


processed_fingerprints = make_fingerprints(reconstructed)
combined_fingerprints = make_fingerprints(combined)

# Count duplicate records as well as unique records
processed_counts = processed_fingerprints.value_counts()
combined_counts = combined_fingerprints.value_counts()

all_records = processed_counts.index.union(
    combined_counts.index
)

processed_counts = processed_counts.reindex(
    all_records,
    fill_value=0
)

combined_counts = combined_counts.reindex(
    all_records,
    fill_value=0
)

same_records = processed_counts.equals(combined_counts)

mismatched_records = int(
    (processed_counts != combined_counts).sum()
)

# ------------------------------------------------
# 8. Verify target labels and row counts
# ------------------------------------------------

target_labels_match = (
    set(reconstructed[target].dropna())
    == set(combined[target].dropna())
)

row_counts_match = len(reconstructed) == len(combined)

print("\n" + "=" * 65)
print("SCALE AND DATASET LINEAGE VERIFICATION")
print("=" * 65)

print("Processed row count:", len(reconstructed))
print("Combined train/test row count:", len(combined))
print("Row counts match:", row_counts_match)

print(
    "All complete records match after transformation:",
    same_records
)

print(
    "Number of record types with mismatched counts:",
    mismatched_records
)

print("Target label sets match:", target_labels_match)

# ------------------------------------------------
# 9. Check transformed feature range
# ------------------------------------------------

print("\n" + "=" * 65)
print("TRANSFORMED FEATURE RANGE")
print("=" * 65)

transformed_min = reconstructed[core_features].min().min()
transformed_max = reconstructed[core_features].max().max()

print("Minimum:", round(transformed_min, 6))
print("Maximum:", round(transformed_max, 6))

# ------------------------------------------------
# 10. Final verification summary
# ------------------------------------------------

print("\n" + "=" * 65)
print("FINAL VERIFICATION SUMMARY")
print("=" * 65)

if (
    row_counts_match
    and same_records
    and target_labels_match
):
    print(
        "SUCCESS: The reconstructed processed dataset matches "
        "the combined training and testing records."
    )
    print(
        "The candidate reverse transformation is consistent "
        "with the observed dataset records."
    )
else:
    print(
        "VERIFICATION NOT PASSED: The datasets do not fully match."
    )
    print(
        "Investigate transformation precision, dataset versions, "
        "and row-level differences before proceeding."
    )

if (
    transformed_min >= 1 - 1e-6
    and transformed_max <= 10 + 1e-6
):
    print("Scale check: Transformed features are within 1–10.")
else:
    print(
        "Scale check: Some transformed values are outside 1–10. "
        "Investigate before using the reconstructed data."
    )

print("\nStep 10F completed.")
```

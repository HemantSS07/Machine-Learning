```python
from pathlib import Path
import pandas as pd
import json

# Search the current notebook folder and its subfolders
BASE_DIR = Path.cwd()

def find_file(filename):
    matches = list(BASE_DIR.rglob(filename))
    if matches:
        return matches[0]
    return None

QUESTIONNAIRE_PATH = find_file("AI_Career_Guidance_Questions.xlsx")
DATASET_PATH = find_file("career_guidance_processed.csv")
DICTIONARY_PATH = find_file("data_dictionary.json")

print("Notebook working directory:", BASE_DIR)
print("Questionnaire found:", QUESTIONNAIRE_PATH)
print("Processed dataset found:", DATASET_PATH)
print("Data dictionary found:", DICTIONARY_PATH)

# Stop if either essential file is missing
if QUESTIONNAIRE_PATH is None:
    raise FileNotFoundError(
        "AI_Career_Guidance_Questions.xlsx was not found. "
        "Place it in the project folder or update QUESTIONNAIRE_PATH."
    )

if DATASET_PATH is None:
    raise FileNotFoundError(
        "career_guidance_processed.csv was not found. "
        "Place it in the project folder or update DATASET_PATH."
    )

# Load questionnaire and dataset
excel_file = pd.ExcelFile(QUESTIONNAIRE_PATH)
print("\nAvailable questionnaire sheets:", excel_file.sheet_names)

questions = pd.read_excel(QUESTIONNAIRE_PATH, sheet_name=excel_file.sheet_names[0])
dataset = pd.read_csv(DATASET_PATH)

print("\nQuestionnaire loaded successfully.")
print("Questionnaire shape:", questions.shape)

print("\nDataset loaded successfully.")
print("Dataset shape:", dataset.shape)
```
